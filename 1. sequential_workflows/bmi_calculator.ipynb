```python
from langgraph.graph import StateGraph, START,END
from typing import TypedDict
```

```python
# state structure for the BMI calculator
class BMIState(TypedDict):
    weight_kg: float
    height_m: float
    bmi: float
    bmi_category: str
```

```python
# define the functions for the calculation of BMI
def calculate_bmi(state: BMIState) -> BMIState:
    weight_kg = state["weight_kg"]
    height_m = state["height_m"]
    bmi = weight_kg / (height_m ** 2)
    state["bmi"] = bmi
    return state

# define the function to categorize BMI
def categorize_bmi(state: BMIState) -> BMIState:
    bmi = state["bmi"]
    if bmi < 18.5:
        state["bmi_category"] = "Underweight"
    elif 18.5 <= bmi < 24.9:
        state["bmi_category"] = "Normal weight"
    elif 25 <= bmi < 29.9:
        state["bmi_category"] = "Overweight"
    else:
        state["bmi_category"] = "Obese"
    return state
```

```python
# create a state graph for the BMI calculator
bmi_graph = StateGraph(BMIState)

bmi_graph.add_node("calculate_bmi", calculate_bmi)
bmi_graph.add_node("categorize_bmi", categorize_bmi)

bmi_graph.add_edge(START, "calculate_bmi")
bmi_graph.add_edge("calculate_bmi", "categorize_bmi")
bmi_graph.add_edge("categorize_bmi", END)

# compile the state graph into a workflow
workflow = bmi_graph.compile()
```

```python
initial_state = {
    "weight_kg": 77.6,
    "height_m": 1.61
}
# invoke the workflow with the initial state
final_state =  workflow.invoke(initial_state)

print(final_state)
```
